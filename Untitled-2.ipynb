```python
# EJERCICIO : TRY / EXCEPT / ELSE / FINALLY

def dividir(a, b):
    try:
        resultado = a / b

    except ZeroDivisionError as error:
        print("EXCEPT: Error, no se puede dividir entre cero.")
        print("Detalle del error:", error)

    else:
        print("ELSE: La operación se realizó correctamente.")
        print("Resultado:", resultado)

    finally:
        print("FINALLY: La operación ha finalizado.\n")


print("CASO 1: División válida")
dividir(20, 4)

print("CASO 2: División entre cero")
dividir(20, 0)
```
