```python
##hacemos la importacion de algunas bibliotecas para la lectura del documento csv
import csv
import os
import numpy as np
import pandas as pd

tabla=pd.read_csv("Height of Male and Female by Country 2022.csv")
tabla.sample(5)


cwd=os.getcwd()
print(cwd)
#resultado c:\Users\Alejandro\Documents
file=open(r'./Height of Male and Female by Country 2022.csv')
csvreader=csv.reader(file)
```

```python
#verificamos los encabezados
header=[]
header=next(csvreader)
header
```

```python
#agregamos los datos al array rows
#hacemos la lectura de las primeras 50 lineas
rows=[]
for row in csvreader:
    rows.append(row)

rows[:50]
```

```python
#sacamos la media, minimo y maximo de las alturas de hombres y mujeres
import numpy as np
dato=np.array(rows)
altura_hombre = dato[1:,2].astype(float)
media_altura_hombre = np.mean(altura_hombre)
minimo_altura_hombre=np.min(altura_hombre)
maximo_altura_hombre=np.max(altura_hombre)
altura_mujer = dato[1:,3].astype(float)
media_altura_mujer=np.mean(altura_mujer)
minimo_altura_mujer=np.min(altura_mujer)
maximo_altura_mujer=np.max(altura_mujer)
print (f"la altura media en hombres es : {media_altura_hombre:.2f}")
print (f"la altura minima en hombres es : {minimo_altura_hombre:.2f}")
print (f"la altura maxima en hombres es : {maximo_altura_hombre:.2f}")
print (f"la altura media en mujeres es :{media_altura_mujer:.2f}")
print (f"la altura minima en mujeres es : {minimo_altura_mujer:.2f}")
print (f"la altura maxima en mujeres es :{maximo_altura_mujer:.2f}")
```
